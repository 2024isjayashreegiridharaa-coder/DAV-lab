```python
import pandas as pd
import numpy as np

# 1. Loading the dataset
# Option A: from a CSV file
# df = pd.read_csv("data.csv")

# Option B: self-contained sample data (with deliberate missing values and a duplicate)
data = {
    "StudentID": [101, 102, 103, 104, 105, 105, 106, 107],
    "Name":      ["Asha", "Ravi", "Meena", "Kiran", "Sneha", "Sneha", "Arjun", "Divya"],
    "Age":       [20, 21, np.nan, 22, 20, 20, 23, np.nan],
    "Marks":     [78, 85, 62, np.nan, 91, 91, 70, 88],
    "Dept":      ["CSE", "ISE", "CSE", "ECE", "ISE", "ISE", None, "CSE"],
}
df = pd.DataFrame(data)

# 2. Basic exploration
print("First 5 rows:\n", df.head())
print("\nShape (rows, columns):", df.shape)
print("\nColumn data types:\n", df.dtypes)
print("\nInfo:")
df.info()

# 3. Descriptive statistics
print("\nDescriptive statistics (numeric):\n", df.describe())
print("\nDescriptive statistics (all columns):\n", df.describe(include="all"))
print("\nMean Marks  :", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Mode Dept   :", df["Dept"].mode()[0])
print("Std Dev Marks:", df["Marks"].std())

# 4. Missing values
print("\nMissing values per column:\n", df.isnull().sum())
print("\nMissing value percentage:\n", (df.isnull().sum() / len(df)) * 100)
print("\nRows containing missing values:\n", df[df.isnull().any(axis=1)])

# 5. Duplicates
print("\nNumber of duplicate rows:", df.duplicated().sum())
print("\nDuplicate rows:\n", df[df.duplicated(keep=False)])

# 6. (Optional) Removing duplicates
df_clean = df.drop_duplicates()
print("\nShape after removing duplicates:", df_clean.shape)
```
